```python
merged_df = pd.merge(df1, df2, on='City')
display(merged_df.head())
```

```python
import numpy as np

def extract_months(time_str):
    if pd.isna(time_str):
        return []
    months = []
    time_str = time_str.replace(' ', '')
    time_parts = time_str.split(',')
    for part in time_parts:
        if '-' in part:
            start_month, end_month = part.split('-')
            months.append(start_month)
            months.append(end_month)
        else:
            months.append(part)
    return months

merged_df['Months'] = merged_df['Best_time_to_visit'].apply(extract_months)
display(merged_df.head())
```

```python
best_cities_by_month = {}

for index, row in merged_df.iterrows():
    city = row['City']
    months = row['Months']
    for month in months:
        if month not in best_cities_by_month:
            best_cities_by_month[month] = []
        if city not in best_cities_by_month[month]:
            best_cities_by_month[month].append(city)

print(best_cities_by_month)
```

```python
print(best_cities_by_month.keys())

def recommend_by_month(month):
    if month in best_cities_by_month:
        return f"Cities recommended for {month}: {', '.join(best_cities_by_month[month])}"
    else:
        return f"No recommendations available for {month}. Please enter a valid month."

month_input = input("Enter the month for which you want recommendations (e.g., October, June): ")
# Based on the keys, convert the input to match the format
month_input_formatted = month_input.title() # Start with title, adjust if needed after seeing keys
recommendations = recommend_by_month(month_input_formatted)
print(recommendations)
```

```python
def extract_duration(duration_str):
    if pd.isna(duration_str):
        return (np.nan, np.nan)
    parts = str(duration_str).split('-')
    if len(parts) == 1:
        duration = int(parts[0])
        return (duration, duration)
    elif len(parts) == 2:
        return (int(parts[0]), int(parts[1]))
    else:
        return (np.nan, np.nan)

merged_df[['Min_duration', 'Max_duration']] = merged_df['Ideal_duration'].apply(extract_duration).tolist()

display(merged_df.head())
```

```python
def recommend_by_month_and_duration(month, duration):
    recommended_cities = []
    for index, row in merged_df.iterrows():
        months = row['Months']
        min_duration = row['Min_duration']
        max_duration = row['Max_duration']
        city = row['City']

        if month in months:
            if not pd.isna(min_duration) and not pd.isna(max_duration):
                if min_duration <= duration <= max_duration:
                    if city not in recommended_cities:
                        recommended_cities.append(city)
            elif pd.isna(min_duration) and pd.isna(max_duration):
                # Handle cases where duration is not specified, maybe recommend if month matches?
                # For now, we'll skip if duration info is missing
                pass


    if recommended_cities:
        return f"Cities recommended for {month} with a duration of {duration} days: {', '.join(recommended_cities)}"
    else:
        return f"No recommendations found for {month} with a duration of {duration} days."

# Example usage (optional, for testing)
# print(recommend_by_month_and_duration('October', 3))
# print(recommend_by_month_and_duration('March', 5))
# print(recommend_by_month_and_duration('July', 2))
```

```python
def recommend_by_month_and_duration(month, duration, sort_by_rating=True):
    recommended_cities_data = []
    for index, row in merged_df.iterrows():
        months = row['Months']
        min_duration = row['Min_duration']
        max_duration = row['Max_duration']
        city = row['City']
        rating = row['Ratings_x']

        if month in months:
            if not pd.isna(min_duration) and not pd.isna(max_duration):
                if min_duration <= duration <= max_duration:
                    recommended_cities_data.append({'City': city, 'Rating': rating})
            elif pd.isna(min_duration) and pd.isna(max_duration):
                pass # Skipping cities with missing duration data

    # Remove duplicate cities while keeping the highest rating if duplicates exist
    unique_cities = {}
    for city_data in recommended_cities_data:
        city = city_data['City']
        rating = city_data['Rating']
        if city not in unique_cities or unique_cities[city]['Rating'] < rating:
            unique_cities[city] = city_data

    recommended_cities_list = list(unique_cities.values())

    if sort_by_rating:
        recommended_cities_list.sort(key=lambda x: x['Rating'], reverse=True)

    if recommended_cities_list:
        city_strings = [f"{city['City']} ({city['Rating']})" for city in recommended_cities_list]
        return f"Cities recommended for {month} with a duration of {duration} days (sorted by rating): {', '.join(city_strings)}"
    else:
        return f"No recommendations found for {month} with a duration of {duration} days."

recommendations = recommend_by_month_and_duration(month_input, duration_input)
print(recommendations)
```
